# Retrieving and Preparing Stock Data with yfinance

In this step, I use the `yfinance` library to extract historical stock data from the Yahoo Finance API. 

The data includes daily prices for major semiconductor companies from 2015 to 2023. 

This process involves data extraction, cleaning, and preparing time series data for analysis. 

The cleaned dataset will be used for trend analysis, financial metrics calculation, and data visualisation in later steps.

Importing `yfinance` and `pandas` libraries.

```python
import yfinance as yf
import pandas as pd
```

Creating a list of semiconductor stock tickers (NDVA, INTC, AMD, QCOM, TSM).

```python
tickers = ['NVDA', 'INTC', 'AMD', 'QCOM', 'TSM']
```

Setting the time period for historical data.

```python
start_date = '2015-01-01'
end_date = '2023-12-31'
```

Creating an emty dictionary to hold data for each ticker.

```python
stock_data = {}
```

Fetching data for each ticker.

```python
for ticker in tickers:
    print("Fetching data for {}...".format(ticker), flush=True)
    stock_data[ticker] = yf.download(ticker, start=start_date, end=end_date, interval="1d",  auto_adjust=False, progress=False )
```

Let's print NVDA prices just to make sure it looks ok.

```python
print(stock_data['NVDA'].head())
print(type(stock_data['NVDA']))
print(stock_data['NVDA'].columns)
```

Data looks ok. So let's combine data into a single DataFrame using adjusted close prices and aligning by date. And print the head.

```python
combined_data = pd.concat([stock_data[ticker] [('Adj Close', ticker)].rename(ticker) for ticker in tickers], axis=1)
print(combined_data.head())
```

Save to CSV

```python
combined_data.to_csv('semiconductor_stock_data.csv')
print("Data fetched and saved to 'semiconductor_stock_data.csv'")
```
